# Amazon Bedrock Model Invocation Methods

Amazon Bedrock provides four main methods for invoking foundation models through Boto3.

| Method | Request Format | Response |
|---|---|---|
| `invoke_model()` | Model-specific | Complete response |
| `invoke_model_with_response_stream()` | Model-specific | Streaming response |
| `converse()` | Unified message format | Complete response |
| `converse_stream()` | Unified message format | Streaming response |

## Key Differences

- **InvokeModel:** Uses the model's specific request and response format.
- **Converse:** Uses a standardized message format across supported models.
- **Streaming:** Returns response events incrementally instead of waiting for the complete response.

## When to Use Each Method

- Use `invoke_model()` when you need direct access to a model-specific API format.
- Use `invoke_model_with_response_stream()` when you need model-specific requests with streaming output.
- Use `converse()` for a consistent conversational interface.
- Use `converse_stream()` for a consistent conversational interface with streaming output.

````python
import boto3
from botocore.exceptions import ClientError, BotoCoreError

# Configuration
AWS_REGION = "us-east-1"
MODEL_ID = "amazon.nova-lite-v1:0"

# Create the Bedrock Runtime client
bedrock_client= boto3.client("bedrock-runtime", region_name=AWS_REGION)
# Request configuration for InvokeModel method only
accept = "application/json"
contentType = "application/json"
# System prompt
system_prompt = """
    You are a helpful AI assistant.
    answer user question in three sentence only
"""
sys_prompt = [
        {"text": system_prompt}
        ]
print(MODEL_ID)
````

## 1. invoke_model()

**Method:** `bedrock_runtime.invoke_model()`

**How it works:** Sends a model-specific JSON request and returns the complete response.

**Usage:** Serialize the request with `json.dumps()`, send it through `invoke_model()`, then parse the response body with `json.loads()`.

````python
import json

user_messages = "What is an API?"

body = {
        
        "messages": [{'role':'user', 'content':[{'text':user_messages}]}],
        "system": sys_prompt,
        "inferenceConfig":{
            "maxTokens": 300,
            "temperature": 0.1,
            "topP": 0.1
        },
    }


try :
    response = bedrock_client.invoke_model(
        modelId=MODEL_ID,
        body=json.dumps(body),
        accept=accept,
        contentType=contentType,
        
    )
    print("response -----------------")
    print(response )
    print("response body -----------------")
    response_body = json.loads(response["body"].read())
    print(response_body)
    answer = response_body["output"]["message"]["content"][0]["text"]
    print("Response from Bedrock model:")
    print(answer)

except (ClientError, BotoCoreError, KeyError, ValueError) as error:
    print(f"Request failed: {error}")
````
